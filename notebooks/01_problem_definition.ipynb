# 01. Project Overview: PlantVillage Disease Classification

Dự án này nghiên cứu bài toán **phân loại bệnh cây từ ảnh lá** trên bộ dữ liệu PlantVillage theo hướng **deep learning, class imbalance handling, transfer learning và lightweight CNN architecture**.

Sau khi hoàn thành 10 phase, trọng tâm nghiên cứu không còn chỉ là “model nào Accuracy cao nhất”, mà là:

- model có nhận diện tốt **minority disease classes** hay không,
- chiến lược xử lý mất cân bằng nào hiệu quả nhất,
- transfer learning có tạo cải thiện đáng kể không,
- kiến trúc MobileNetV2 cải tiến có đáng dùng so với baseline không,
- mô hình cuối có phù hợp cho deployment nhẹ trên edge/mobile không.


# 02. Problem Definition

Bài toán được định nghĩa là **multi-class image classification**.

| Thành phần | Mô tả |
|---|---|
| Input | Ảnh lá cây từ PlantVillage |
| Output | Nhãn class gồm cây trồng và trạng thái bệnh |
| Số class | 38 class |
| Dữ liệu chính | PlantVillage color images |
| Vấn đề chính | Severe class imbalance giữa majority và minority classes |
| Mục tiêu ML | Tối ưu Macro F1, Minority Recall và Worst-class Recall, không chỉ Accuracy |

Trong bối cảnh bệnh cây, **False Negative ở class bệnh hiếm** rất quan trọng: nếu cây thật sự bị bệnh nhưng model dự đoán sai sang class khỏe hoặc class khác, hệ thống có thể bỏ sót cảnh báo.


# 03. Dataset Context

Dự án bắt đầu bằng việc audit các phiên bản dữ liệu PlantVillage:

- `color`: ảnh RGB gốc.
- `grayscale`: ảnh xám.
- `segmented`: ảnh đã tách nền.

Sau các phase baseline và preprocessing, hướng nghiên cứu chính tập trung vào **color dataset**, vì ảnh màu giữ lại thông tin màu sắc bệnh học quan trọng như đốm vàng, nâu, rỉ sắt, cháy lá và biến đổi texture.

Các split train/validation/test được giữ cố định qua các phase sau để tránh data leakage và đảm bảo so sánh công bằng.


# 04. Research Motivation

Dự án được thúc đẩy bởi bốn vấn đề nghiên cứu chính:

1. **Class imbalance nghiêm trọng**  
   Một số bệnh/cây có rất nhiều ảnh, trong khi một số class hiếm có rất ít mẫu. Accuracy có thể cao dù model nhận diện kém class thiểu số.

2. **Minority recognition và fairness**  
   Trong ứng dụng bệnh cây, model cần hoạt động ổn định trên cả class phổ biến và class hiếm. Vì vậy Macro F1, Minority Recall và Worst-class Recall quan trọng hơn Accuracy đơn lẻ.

3. **Transfer learning**  
   SimpleCNN học từ đầu có giới hạn. MobileNetV2 pretrained có thể tận dụng ImageNet features để học tốt hơn texture và visual patterns của bệnh lá.

4. **Lightweight deployment**  
   Mô hình cuối nên đủ nhẹ để phù hợp với edge/mobile deployment, ví dụ điện thoại hoặc thiết bị hỗ trợ nông nghiệp.


# 05. Research Questions Sau Khi Hoàn Thành 10 Phase

Dự án trả lời các câu hỏi nghiên cứu sau:

| RQ | Câu hỏi nghiên cứu |
|---|---|
| RQ1 | Imbalance handling có cải thiện minority recognition không? |
| RQ2 | Loss-level balancing có vượt sampler-based balancing không? |
| RQ3 | Hybrid strategy giữa best sampler và best loss có vượt các hướng đơn lẻ không? |
| RQ4 | Hai mô hình học sâu hiện đại MobileNetV2 và EfficientNet-B0 mô hình nào phù hợp hơn? |
| RQ5 | Improved MobileNetV2 với residual branch/attention có outperform MobileNetV2 baseline không? |
| RQ6 | Mô hình nào đạt cân bằng tốt nhất giữa performance, fairness và efficiency? |


# 06. Hypotheses

Các giả thuyết nghiên cứu chính:

- **H1:** WeightedRandomSampler giúp tăng exposure của minority classes và cải thiện Minority Recall.
- **H2:** Loss-level balancing như Weighted CE, Focal Loss và Class-Balanced Loss có thể cải thiện class khó, nhưng không nhất thiết vượt sampler-based balancing.
- **H3:** Phase 07 kết hợp winner của Phase 05 và Phase 06 để kiểm tra hybrid SimpleCNN strategy.
- **H4:** MobileNetV2/EfficientNet-B0 pretrained outperform SimpleCNN nhờ tận dụng ImageNet features.
- **H5:** Residual enhancement branch giúp học local disease textures tốt hơn MobileNetV2 baseline.
- **H6:** Attention nhẹ như SE/CBAM giúp model tập trung hơn vào vùng bệnh, nhưng chỉ đáng dùng nếu cải thiện metric tương xứng với chi phí tính toán.

# 08. Experimental Design Logic

Thiết kế thực nghiệm đi theo hướng tăng dần độ phức tạp:

1. **Hiểu dữ liệu**  
   Audit class distribution, image properties, duplicate và imbalance.

2. **Thiết lập baseline**  
   So sánh dataset variants bằng SimpleCNN baseline.

3. **Xử lý imbalance**  
   Nghiên cứu sampler-based và loss-level strategies trên SimpleCNN; Phase 07 kết hợp best sampler/data-level strategy và best loss-level strategy bằng SimpleCNN trước khi sang pretrained backbones.

4. **Nâng cấp backbone**  
   Chuyển từ SimpleCNN sang hai pretrained backbones hiện đại: MobileNetV2 và EfficientNet-B0.

5. **Cải tiến kiến trúc**  
   Thêm residual branch và attention nhẹ để học disease-specific features.

6. **Tổng hợp khoa học**  
   So sánh toàn bộ phase, phân tích fairness, efficiency, explainability, limitation và future work.


# 09. Evaluation Metrics

Các metric chính:

| Metric | Ý nghĩa |
|---|---|
| Accuracy | Tỷ lệ dự đoán đúng tổng thể |
| Macro Precision | Precision trung bình không trọng số trên các class |
| Macro Recall | Recall trung bình không trọng số trên các class |
| Macro F1 | Metric chính vì cân bằng precision/recall theo class |
| Minority Recall | Khả năng nhận diện các class thiểu số |
| Worst-class Recall | Recall của class kém nhất, phản ánh fairness cực trị |
| Per-class Recall/F1 | Phân tích chi tiết từng class |
| Confusion Matrix | Phân tích class bị nhầm lẫn |
| Params / Inference Time | Đánh giá tính nhẹ và khả năng deployment |

**Macro F1**, **Minority Recall** và **Worst-class Recall** là các metric quan trọng nhất trong nghiên cứu này vì dataset bị mất cân bằng.


# 10. Final Proposed Direction

Sau toàn bộ project, hướng mô hình cuối là:

> **Modern pretrained CNN backbone + imbalance-aware training**

Trong đó:

- MobileNetV2/EfficientNet-B0 pretrained cung cấp backbone mạnh và nhẹ.
- WeightedRandomSampler tăng exposure của minority classes.
- Light augmentation giảm memorization khi minority samples bị lặp.
- EarlyStopping theo Validation Macro F1 kiểm soát overfitting.
- Improved MobileNetV2 với residual branch/attention được đánh giá như phương án cải tiến kiến trúc.

Phase 10 sẽ quyết định final model dựa trên bằng chứng tổng hợp, không chỉ dựa vào một metric đơn lẻ.
